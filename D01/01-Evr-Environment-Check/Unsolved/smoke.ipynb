## 1. The libraries every day uses

The first five lines are the ones the README checks. A different number means the notebook is not running on the `.venv` kernel.

```python
import sys

import matplotlib
import numpy as np
import pandas as pd
import polars as pl
import seaborn as sns
import sklearn

print("python      ", ".".join(map(str, sys.version_info[:3])))
print("scikit-learn", sklearn.__version__)
print("pandas      ", pd.__version__)
print("polars      ", pl.__version__)
print("numpy       ", np.__version__)
print("matplotlib  ", matplotlib.__version__)
print("seaborn     ", sns.__version__)
```

## 2. The libraries D02 and D03 add

Each line prints a version, or `MISSING` if the import fails. On the course VM every line has a version. In the Colab fallback, `MISSING` lines are expected: Colab does not have them, and you are back on your VM before D03 needs them. `MLFLOW_DISABLE_TELEMETRY` stops MLflow from sending usage data.

```python
import importlib
import os

os.environ["MLFLOW_DISABLE_TELEMETRY"] = "true"
for name in ["scipy", "pyarrow", "xgboost", "lightgbm", "shap", "optuna", "mlflow", "joblib", "skops",
             "fastapi", "uvicorn"]:
    try:
        module = importlib.import_module(name)
        print(f"{name:<12} {module.__version__}")
    except ImportError:
        print(f"{name:<12} MISSING")
```
