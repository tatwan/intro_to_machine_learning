# Your first model, end to end

One row is one penguin. The question: given four body measurements, which of three species is it? Follow the README's numbered steps; each step below has its own cell. The README shows the code for steps 1 to 5.

## Step 1 · Load the table

Keep the four measurement columns as `X` and the species as `y`.

## Step 2 · Split once, before anything learns

A quarter of the rows go to the test set; `stratify=y` keeps the same share of each species in both parts.

## Step 3 · Build the pipeline and fit it on the training rows

Fill missing values with the median, scale, then classify.

## Step 4 · Score it, and score a model that knows nothing

The dummy always answers the most common species in the training rows.

## Step 5 · Ask about a penguin the model has never seen

Bill 47.0 mm long and 15.0 mm deep, flipper 215 mm, mass 5,200 g.

## Step 6 · Swap the model, keep everything else

Change only the last step of the pipeline: `DecisionTreeClassifier(random_state=42)`, then `RandomForestClassifier(random_state=42)`, then `KNeighborsClassifier()`. The imputer, the scaler and the split stay as they are. Write down each test accuracy.
